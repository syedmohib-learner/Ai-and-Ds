# Mini Inventory Management System

#### Task 1

```python
store_name = input("Enter Store Name:")
manager_name = input("Enter Manager Name:")
store_code = "ELEC-2026-KHI"
Category_code = store_code[0:SM4]
Year = store_code[5:9]
City_code = store_code[10:]
print(f"""The Category Code is: {Category_code}
The Year is: {Year}
The City Code is: {City_code}""")
```

#### Task 2

```python
products = ["Laptop", "Mouse", "Keyboard", "Monitor", "Headphones"]
print(products)
len(products)
```

#### Task 3

```python
user_input = input("Enter Product Name:")
products.append(user_input)
new_product =  ["Webcam", "USB Hub", "Speaker"]
products.extend(new_product)
```

#### Task 4

```python
monitor_position = products.index("Monitor")
products.insert(monitor_position + 1, "Printer")
```

#### Task 5

```python
products.remove("Headphones")
```

#### Task 6

```python
shipment_list = ["Tablet", "Smart Watch", "Power Bank"]
products.extend(shipment_list)
removed_product = products.pop()
print(f"Removed Product: {removed_product}")
```

#### Task 7

```python
print(products.index("Laptop"))
products.count("Mouse")
```

#### Task 8

```python
sorted_asc = products.copy()
sorted_asc.sort()
sorted_desc = products.copy()
sorted_desc.sort(reverse=True)
print(f"From A-Z: {sorted_asc}")
print(f"From Z-A: {sorted_desc}")
```

#### Task 9

```python
products.reverse()
print(f"Reversed Inventory: {products}")
```

#### Task 10

```python
backup_products = products.copy()
print(f"""Current Inventory: {products}
Backup Inventory: {backup_products}""")
```

#### Task 11

```python
temporary_list = ["Test Laptop", "Test Mouse", "Test Keyboard"]
temporary_list.clear()
print(f"Cleared temporary list: {temporary_list}")
```

#### Task 12

```python
featured_products = products[1:4]
print(f"Featured Products: {featured_products}")
```

```python
print("=" * 50)
print("        MINI INVENTORY MANAGEMENT REPORT        ")
print("=" * 50)
# Store Information
print("\n--- STORE INFORMATION ---")
print(f"Store Name   : {store_name}")
print(f"Manager Name : {manager_name}")
print(f"Store Code   : {store_code}")
print(f"Category Code: {Category_code}")
print(f"Year         : {Year}")
print(f"City Code    : {City_code}")
# Inventory Summary
print("\n--- INVENTORY SUMMARY ---")
print(f"Total Products Count : {len(products)}")
print(f"Current Inventory    : {products}")
# Results of Operations
print("\n--- INVENTORY OPERATIONS ---")
print(f"Removed Product (pop): {removed_product}")
# Sorted & Reversed Inventory
print("\n--- SORTED & REVERSED INVENTORY ---")
print(f"Sorted Inventory A-Z : {sorted(products)}")
print(f"Sorted Inventory Z-A : {sorted(products, reverse=True)}")
# Backup & Featured Products
print("\n--- BACKUP & FEATURED PRODUCTS ---")
print(f"Backup Inventory     : {backup_products}")
print(f"Featured Products    : {featured_products}")
# Cleared Temporary Inventory
print("\n--- TEMPORARY INVENTORY ---")
print(f"Cleared Temp List    : {temporary_list}")
print("\n" + "=" * 50)
```
