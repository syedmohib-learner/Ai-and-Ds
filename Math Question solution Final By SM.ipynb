```python
# Area of a Rectangle: Write a Python program to calculate the area of a rectangle. Use the formula: 𝐴𝒓𝒆𝒂 = 𝒍𝒆𝒏𝒈𝒕𝒉 × 𝒘𝒊𝒅𝒕𝒉 Take length and width as inputs from the user. 
length= int(input("Enter length:"))
width= int(input("Enter width:"))
area= length*width
print(f"The Area of Rectangle is: {area}")
```

```python
#Circumference of a Circle: Write a Python program to calculate the circumference of a circle. Use the formula: 𝑪𝒊𝒓𝒄𝒖𝒎𝒇𝒆𝒓𝒆𝒏𝒄𝒆 = 𝟐𝝅𝒓 Take the radius r as input from the user. 
radius=float(input("Enter radius"))
pi= 3.142
circumference= 2 * pi * radius
print(f"The Circumference of Circle is: {circumference:}")
```

```python
#Simple Interest: Write a Python program to calculate the simple interest. Use the formula: 𝑺𝒊𝒎𝒑𝒍𝒆 𝑰𝒏𝒕𝒆𝒓𝒆𝒔𝒕 = 𝑷𝒓𝒊𝒏𝒄𝒊𝒑𝒂𝒍 × 𝑹𝒂𝒕𝒆 × 𝑻𝒊𝒎𝒆  Take Principal, Rate, and Time as inputs from the user. 
principal=int(input("Enter Principal:"))
rate=int(input("Enter Rate (in %):"))
time=int(input("Enter Time(in years) :"))
simple_interest= principal*rate*time
print(f"The Simple Interest is: {simple_interest}")
```

```python
#Speed of an Object: Write a Python program to calculate the speed of an object. Use the formula:  𝑆𝒑𝒆𝒆𝒅     =      𝑫𝒊𝒔𝒕𝒂𝒏𝒄𝒆 /    𝑻𝒊𝒎𝒆  Take Distance and Time as inputs from the user.
distance=int(input("Enter Dstance (in m) :"))
time=int(input("Enter Time (in second):"))
speed=round(distance/time,2)
print(f"The speed of an Object is: {speed}")
```

```python
## Write a Python program to calculate compound interest. Use the formula:  
p=int(input("Enter Principal Amount is:"))
r=float(input("Enter Annual Interest Rate (in decimal):"))
n=int(input("Enter Number of Time Interest is compounded per year:"))
t=int(input("Enter Time in Years:"))
A= round(p*((1+(r/n))**(n-t)),2)
compound_interest= A - p
print(f"The Amount is: {A}")
print(f"The Compound Interest is: {compound_interest}")
```

```python
#Write a Python program to calculate the kinetic energy of an object. Use the formula: 𝑲𝑬=𝟏 𝟐 𝒎 𝑽𝟐 Take m (mass in kilograms) and v (velocity in meters/second) as inputs from the user.
m=float(input("Enter mass in kilogram"))
v=float(input("Enter velocity in meter"))
KE = round(0.5 * m * v ** 2,2)
print(f"The Kinetic energy of an object is: {KE}")
```

```python
#Quadratic Equation Roots: Write a Python program to find the roots of a quadratic equation. Use the formula: X =  ±√ Take a, b, and c as inputs from the user. 
a = float(input("Enter coefficient a: "))
b = float(input("Enter coefficient b: "))
c = float(input("Enter coefficient c: "))
x1 = (-b + (b**2 - 4 * a * c) ** 0.5) / (2 * a)
x2 = (-b - (b**2 - 4 * a * c) ** 0.5) / (2 * a)
print("First root (x1):", x1)
print("Second root (x2):", x2)
```

```python
# Temperature Conversion: Write a Python program to convert a temperature from Celsius to Fahrenheit. Use the formula:  Take C (temperature in Celsius) as input from the user
C=float(input("Enter C:"))
F=round((9/5)*C+32,2)
print(f"The Temperature in Fahrenheit is: {F}")
```

```python
#Gravitational Force: Write a Python program to calculate the gravitational force between two objects. Use the formula 𝑭 =𝑮𝒎𝟏∗ 𝒎𝟐 𝒓𝟐 m1, m2 are the masses of the objects r = distance between the centers of the objects Take m1, m2 and r as inputs from the user.
m1=int(input("Enter m1 (in kgs):"))
m2=int(input("Enter m2 (in kgs):"))
r=int(input("Enter r:"))
G=6.67e-11
F = G*(m1*m2)/(r**2)
print(f"The Gravitational force is: {F}")
```

```python
# Volume of a Cylinder: Write a Python program to calculate the volume of a cylinder. Use the formula:  𝑽𝒐𝒍𝒖𝒎𝒆 = 𝝅𝒓𝟐𝒉 Take radius(r) and height(h) as inputs from the user
r=int(input("Enter Radius:"))
h=int(input("Enter Height:"))
pi=3.142
Volume= round(pi*(r**2)*h,2)
print(f"The Volume of Cylinder is: {Volume}")
```

```python
#Future Value in Savings: Write a Python program to calculate the future value of an investment. Use the formula:  𝐹𝑉 =𝑃𝑉(1+𝑟) where: o FV= future value o PV= present value o r= annual interest rate (as a decimal) o t= time in years Take PV, r, and t as inputs from the user. 
PV=int(input("Enter Present Value:"))
r=float(input("Enter Annual Interest Rate:"))
t=int(input("Enter Time in Year:"))
FV=round(PV*((1+r)**t),2)
print(f"The Future Value of an Investment is: {FV}")
```

```python
#Work Done by a Force: Write a Python program to calculate the work done by a force. Use the formula:   where: W = work done f= force d = distance theta = angle between force and direction of movement (in degrees) Take f, d, and θ as inputs from the user.
f = float(input("Enter force (f) in Newtons: "))
d = float(input("Enter distance (d) in meters: "))
cos_theta = float(input("Enter cos(θ) value (e.g. 1 for 0°, 0.5 for 60°): "))
W = f * d * cos_theta
print(f"Work Done (W) = {W} Joules")
```

```python
# Heat Transfer: Write a Python program to calculate the amount of heat transferred. Use the formula:  𝑄 =𝑚𝑐 (∆T) where: Q= heat transfer m = mass c = specific heat capacity ΔT Take m, c, T as inputs from the user. 
m=float(input("Enter Mass"))
c=int(input("Enter Specific Heat Capacity"))
T1=int(input("Enter Temperature 1 in Celsius"))
T2=int(input("Enter Temperature 2 in Celsius"))
delta_t=T2-T1
Q=m*c*delta_t
print(f"The Amount of Heat transfered is: {Q}")
```
