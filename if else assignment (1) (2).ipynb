# 1. Check Discount Eligibility
Write a program to check if a customer is eligible for a discount. If the total purchase is
more than $100, apply a 10% discount and display the final price. Otherwise, display the
total price as it is

```python
total_purchase = int(input("Enter your Total Purchase $"))
if total_purchase > 100:
    discount = total_purchase * 0.1
    price_after_discount = total_purchase - discount
    print("After 10% discount, you have to pay $",price_after_discount)
else:
    print("You have to pay $",total_purchase)
```

# 3. Membership Discount
Check if the customer is a member (is_member = True). Members get a 20% discount;
non-members get a 5% discount. Calculate and print the discounted price.

```python
amount = int(input("Enter your amount"))
member = input("Are you member? True/False").capitalize()
if member == "True":
    discount = amount * 0.2
    discounted_price = amount - discount
    print("You are the member ,you will get 20% discount, you have to pay $",discounted_price)
else:
    discount = amount * 0.05
    discounted_price = amount - discount
    print("You are not the member ,you will get 5% discount, you have to pay $",discounted_price)
```

# Buy-One-Get-One-Free
If a customer buys an even number of items, they get half of them for free. Otherwise,
they pay for all. Calculate the number of items the customer has to pay for.

```python
no_of_item = int(input("Enter how many items you buy?"))
if no_of_item % 2 == 0:
    item_to_pay = no_of_item // 2
    print("You get half for free! You only have to pay for" ,item_to_pay,"items.")
else:
    item_to_pay = no_of_item
    print("No offer applies. You have to pay for all",item_to_pay ,"items.")    
```

# Sales Tax
If the price of an item is greater than $500, apply a luxury tax of 15%. Otherwise, apply a
standard tax of 8%. Display the total price after tax.

```python
price = int(input("Enter your Price $:"))
if price > 500:
    tax = price * 0.15
    total_tax = price + tax
    print("Total price after 15% tax $",total_tax)
else:
    tax = price * 0.08
    total_tax = price + tax
    print("Total price after 8% tax $",total_tax)
```

# Tiered Discounts
Apply discounts based on the total price:
$0–$50: No discount.
$50–$100: 10% discount.
Over $100: 20% discount.

```python
total_price = int(input("Enter Price: $"))
if total_price >= 0 and total_price < 50:
    print("You get No discount")
elif total_price >= 50 and total_price < 100:
    discount = total_price * 0.1
    discounted_price = total_price - discount
    print("10% discount applied, pay $",discounted_price)
else:
    discount = total_price * 0.2
    discounted_price = total_price - discount
    print("20% discount applied, pay $",discounted_price)     
```
