7.Write a program that prompts the user to enter their weight (in kilograms) and height
(in meters). The program should calculate the Body Mass Index (BMI) using the formula:
BMI = weight / (height * height). The program should then classify the BMI into one of
the following categories:
less than 18.5 - Underweight
BMI between 18.5 and 24.9 - Normal weight
BMI between 25 and 29.9 - Overweight
BMI 30 or greater - Obesity 

```python
weight = float(input("Enter Weight (in kilogram) :"))
height = float(input("Enter Height (in meters) :"))
BMI = round(weight / (height * height),2)
print("The Body Mass Index (BMI) is",BMI)
if BMI < 18.5:
    print("Underweight")
elif BMI >= 18.5 and BMI <= 24.9:
    print("Normal weight")
elif BMI >25  and BMI <= 29.9:
    print("Overweight")
else:
    print("Obesity")
```

9. The roots of the quadratic equation ax2
 + bx + c = 0, a ≠ 0 are given by the following
formula:
In this formula, the term b2
 - 4ac is called the discriminant. If b2
 - 4ac = 0, then the
equation has two equal roots.
If b2
 - 4ac > 0, the equation has two real roots. If b2
 - 4ac < 0, the equation has two
complex roots.
Write a program that prompts the user to input the value of a (the coefficient of x2
), b
(the coefficient of x), and c (the constant term) and outputs the roots of the quadratic
equation. 

```python
a = int(input("Enter the coefficient of x2: "))
b = int(input("Enter the coefficient of x: "))
c = int(input("Enter the constant term: "))
discriminant = b**2 - 4 * a * c
if discriminant > 0:
    root1 = (-b + discriminant**0.5) / (2 * a)
    root2 = (-b - discriminant**0.5) / (2 * a)
    print("Two real roots:", root1, "and", root2)
elif discriminant == 0:
    root = -b / (2 * a)
    print("Two equal roots:", root)
else:
    real_part = -b / (2 * a)
    imag_part = (abs(discriminant)**0.5) / (2 * a)    
    root1 = complex(real_part, imag_part)
    root2 = complex(real_part, -imag_part)
    print("Two complex roots:", root1, "and", root2)
```

ascending order. The program should print the sorted numbers.

13. Write a program that prompts the user to input a year and determine whether the
year is a leap year or not.
Leap Years are any year that can be evenly divided by 4. A year that is evenly divisible by
100 is a leap year only if it is also evenly divisible by 400. Example:
1992 Leap Year
2000 Leap Year
1900 NOT a Leap Year
1995 NOT a Leap Year 


```python
year = int(input("Enter a year: "))
if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(year," is a leap year.")
else:
    print(year," is not a leap year.")
```

14. Write a program that prompts the user to input number of calls and calculate the
monthly telephone bills as per the following rule:
Minimum Rs. 200 for up to 100 calls.
Plus Rs. 0.60 per call for next 50 calls.
Plus Rs. 0.50 per call for next 50 calls.
Plus Rs. 0.40 per call for any call beyond 200 calls. 

```python
number_of_call = int(input("Enter numbers of calls:"))
price = 200
if number_of_call <= 100:
    charges = price
    print("Your Number of Calls are within 100, You have to pay: Rs",charges)
elif number_of_call > 100 and number_of_call <= 150:
    exceeding = number_of_call - 100
    charged_upon_call = exceeding * 0.60
    charges = price + charged_upon_call
    print("Your Number of Calls are within 150, You have to pay: Rs",charges)
elif number_of_call > 150 and number_of_call <= 200:
    exceeding_one = number_of_call - 150
    charged_upon_call_one = exceeding_one * 0.50
    charges_one = price + 30 + charged_upon_call_one
    print("Your Number of Calls are within 200, You have to pay: Rs",charges_one)
else:
    exceeding_two = number_of_call - 200
    charged_upon_call_two = exceeding_two * 0.40
    charges_two = price + 30 + 25 + charged_upon_call_two
    print("Your Number of Calls are above 200, You have to pay: Rs",charges_two)
```
