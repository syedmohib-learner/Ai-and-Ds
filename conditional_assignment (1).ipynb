## A traffic system calculates fines:

Speed limit = 60 km/h
* If speed ≤ 60 → No fine
* If 61–80 → Fine = 500
* If 81–100 → Fine = 2000
* If >100 → License suspended
#### BUT:
*If emergency vehicle → no fine
*If rainy weather AND speed > 60 → fine increases by 50%

```python
is_emergency = input("Is it an emergency vehicle? (yes/no): ").strip().lower()
if is_emergency == "yes":
    print("No fine")
else:
    speed = float(input("Enter vehicle speed (km/h): "))
    is_rainy = input("Is it rainy weather? (yes/no): ").strip().lower()
    if speed <= 60:
        print("No fine")
    else:
        if speed <= 80:
            base_fine = 500
            if is_rainy == "yes":
                fine = base_fine * 1.5
                print(f"Fine: {fine}")
            else:
                print(f"Fine: {base_fine}")
        elif speed <= 100:
            base_fine = 2000
            if is_rainy == "yes":
                fine = base_fine * 1.5
                print(f"Fine: {fine}")
            else:
                print(f"Fine: {base_fine}")
        else:
            print("License suspended")
```

## A company filters candidates:
* Experience ≥ 5 years AND skill score ≥ 80 → shortlisted
* Experience 2–4 years AND skill score ≥ 70 → interview
* Freshers → only if test score ≥ 85
* If criminal record = True → reject immediately
* If multiple job switches (>3 in 2 years) → flag as unstable

```python
criminal_record = input("Criminal record? (yes/no): ").strip().lower()
if criminal_record == "yes":
    print("Rejected immediately")
else:
    job_switches = int(input("Enter number of job switches in the last 2 years: "))  
    if job_switches > 3:
        print("Flagged as unstable")   
    experience = float(input("Enter years of experience: "))
    if experience >= 5:
        skill_score = float(input("Enter skill score: "))
        if skill_score >= 80:
            print("Shortlisted")
        else:
            print("Not shortlisted")           
    elif 2 <= experience <= 4:
        skill_score = float(input("Enter skill score: "))
        if skill_score >= 70:
            print("Interview")
        else:
            print("Not selected for interview")         
    else: 
        test_score = float(input("Enter test score: "))
        if test_score >= 85:
            print("Eligible")
        else:
            print("Not eligible")
```

## Write a program to decide passenger clearance status.
Take input:
* baggage_weight
* passport_valid (Yes/No)
* visa_status (Valid/Invalid)
* criminal_check (Clear/Flagged)
#### Rules:
* baggage > 30kg → extra charge
* passport invalid → reject
* visa invalid → reject
* criminal flagged → security hold
* otherwise → cleared for boarding

```python
passport_valid = input("Passport valid? (yes/no): ").strip().lower()
visa_status = input("Visa status (valid/invalid): ").strip().lower()
criminal_check = input("Criminal check (clear/flagged): ").strip().lower()
if passport_valid != "yes":
    print("Rejected")
elif visa_status != "valid":
    print("Rejected")
elif criminal_check == "flagged":
    print("Security hold")
else:
    baggage_weight = float(input("Enter baggage weight (kg): "))
    if baggage_weight > 30:
        print("Extra charge applies")
    print("Cleared for boarding")
```

## Write a program to detect suspicious banking transactions.
Take input:
* transaction_amount
* location_match (Yes/No)
* time_of_transaction(Day/Night)
* previous_transactions_count
* device_verified (Yes/No)
#### Rules:
* amount > 100,000 → suspicious
* location mismatch + high amount → fraud alert
* night transaction + unverified device → high risk
* many transactions (>10 in 1 hour) → freeze account
* otherwise → normal transaction

```python
prev_tx_count = int(input("Enter previous transactions count in past 1 hour: "))
if prev_tx_count > 10:
    print("Freeze account")
else:
    amount = float(input("Enter transaction amount: "))
    location_match = input("Location match? (yes/no): ").strip().lower()
    time_of_tx = input("Time of transaction (day/night): ").strip().lower()
    device_verified = input("Device verified? (yes/no): ").strip().lower()
    if location_match == "no" and amount > 100000:
        print("Fraud alert")
    elif time_of_tx == "night" and device_verified == "no":
        print("High risk")
    elif amount > 100000:
        print("Suspicious")
    else:
        print("Normal transaction")
```

## Write a program to classify a person’s risk level in a police database.
Take input:
* criminal_record (Yes/No)
* number_of_cases
* travel_history_flagged (Yes/No)
* financial_activity_score
* last_activity_days
#### Rules:
* criminal_record = Yes AND cases > 3 → High Risk
* flagged travel + suspicious financial score → High Risk
* no activity for long time (>365 days) → Monitor
* clean record + normal activity → Low Risk
* mixed signals → Medium Risk

```python
criminal_record = input("Criminal record? (yes/no): ").strip().lower()
cases = int(input("Enter number of cases: "))
travel_flagged = input("Travel history flagged? (yes/no): ").strip().lower()
financial_score = float(input("Enter financial activity score (0-100): "))
last_activity_days = int(input("Enter days since last activity: "))
is_financial_suspicious = financial_score < 30 or financial_score > 80
if (criminal_record == "yes" and cases > 3) or (travel_flagged == "yes" and is_financial_suspicious):
    print("High Risk")
elif last_activity_days > 365:
    print("Monitor")
elif criminal_record == "no" and travel_flagged == "no" and not is_financial_suspicious:
    print("Low Risk")
else:
    print("Medium Risk")
```
